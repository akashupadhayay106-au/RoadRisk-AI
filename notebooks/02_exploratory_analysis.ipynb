## 2. Temporal Analysis: Monthly Trend Trajectory
Evaluate whether incidents show annual seasonality or monthly volatility.


```python
df_monthly = get_monthly_trends()
plt.figure(figsize=(14, 4))
plt.plot(df_monthly['year_month'], df_monthly['accident_count'], marker='o', color='#FF4B4B', label='Accidents')
plt.plot(df_monthly['year_month'], df_monthly['total_casualties'], marker='s', color='#FF8C42', linestyle='--', label='Casualties')
plt.xticks(rotation=45)
plt.title('Monthly Accident and Casualty Trajectory (2022–2025)')
plt.xlabel('Year-Month')
plt.ylabel('Count')
plt.grid(True, alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()
```

## 3. Crash Hour Density & Peak Hour Impact
Inspect accident frequencies across 24-hour cycles.


```python
hourly_sql = 'SELECT hour, COUNT(*) as count, SUM(casualties) as casualties FROM accidents_india GROUP BY hour ORDER BY hour ASC;'
df_hour = execute_query(hourly_sql)

plt.figure(figsize=(10, 4))
sns.barplot(data=df_hour, x='hour', y='count', palette='Reds_d')
plt.title('Crash Frequency by Hour of Day')
plt.xlabel('Hour (0–23)')
plt.ylabel('Incident Count')
plt.show()
```
